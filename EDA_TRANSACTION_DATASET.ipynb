```python
def parse_csv(filename):
    with open(filename,"r",buffering=8192) as file:
        next(file)
        for line in file:
            line = line.strip()
            if not line:
                continue

            yield line.split(",")
```

```python
for row in parse_csv("transaction_dataset.xls"):
    print(row)
```

```python
def process_file(filename):
    count = 0
    mean = 0
    M2 = 0

    category_frequency = {}

    for row in parse_csv(filename):
        transaction_id = row[0]
        category = row[1]
        amount = float(row[2])

        count += 1

        # Welford's algorithm for mean and variance
        delta = amount - mean
        mean += delta / count

        delta2 = amount - mean
        M2 += delta * delta2

        # Count frequency of each category
        if category in category_frequency:
            category_frequency[category] += 1
        else:
            category_frequency[category] = 1

    variance = M2 / (count - 1) if count > 1 else 0

    return count, mean, variance, category_frequency
```

```python
count, mean, variance, category_frequency = process_file("transaction_dataset.xls")

print("Total Records :", count)
print("Running Mean  :", round(mean, 2))
print("Variance      :", round(variance, 2))

print("\nCategory Frequencies")

for category, freq in category_frequency.items():
    print(category,":", freq)
```
