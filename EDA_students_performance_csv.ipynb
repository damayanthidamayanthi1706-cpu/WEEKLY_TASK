```python
df["Total Score"] = (
    df["math score"] +
    df["reading score"] +
    df["writing score"]
    )
print(df[["math score", "reading score", "writing score","Total Score"]].head())
```

```python
df["Average Score"] = df["Total Score"] / 3

print(df[["Total Score","Average Score"]].head())
```

```python
df["Percentage Performance"] = (df["Total Score"] / 300) * 100

print(df[["Total Score","Percentage Performance"]].head())
```

```python
df["Pass/Fail Status"] = df["Average Score"].apply(
    lambda x: "Pass" if x >= 40 else "Fail"

)
print(df[["Average Score","Pass/Fail Status"]].head())
```

```python
print(df.head())
```

```python
for col in subjects:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print(col)
    print("Lower Limit:", lower)
    print("Upper Limit:", upper)
    print("Number of Outliers:", len(outliers))
    print()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

parental_avg = df.groupby("parental level of education")[
    ["math score", "reading score", "writing score"]
].mean()

print(parental_avg)
```

Writing (~13 points from high school to master's), then Reading (~11) and Math (~8).

Yes. High school and some-high-school groups sit around 63-64, while degree holders score higher.

## Lunch Type

**Q1. How does performance vary?**
Standard lunch students outperform free/reduced lunch students.

**Q2. Which group has the higher average?**
Standard lunch.

**Q3. Are differences consistent across subjects?**
Yes. The gap is largest in Math (~11 points) and ~7-8 points in Reading and Writing.

**Q4. What do the plots show?**
Standard lunch box plots sit higher. The free/reduced group has a lower median and more low outliers.

## Test Preparation

**Q1. How do the two groups compare?**
Students who completed test prep score higher.

**Q2. Which subject shows the largest difference?**
Writing (~10 points), then Reading (~7) and Math (~6).

**Q3. How does Total Score vary?**
About 222 for completed vs ~208 for not completed.

**Q4. How does Average Score vary?**
About 7-8 points higher for the completed group.

**Q5. What does the box plot reveal?**
The completed group has a higher median and fewer low outliers, with similar spread.

**Q6. Can this be interpreted as causation?**
No. This is observational data. Motivation, prior ability and family support could explain the gap.

## Gender and Race/Ethnicity

**Q1. How do scores vary by gender?**
Females lead in Reading (~7 points) and Writing (~9). Males lead in Math (~5).

**Q2. How do scores vary by race/ethnicity?**
Group E is highest and Group A lowest, with B, C and D in between.

**Q3. Which subjects show the largest differences?**
Race/ethnicity: Math (~12 points between groups). Gender: Writing.

**Q4. Which groups may need more support?**
Group A, females in Math, males in Reading and Writing, free/reduced lunch students, and students without test prep.

**Q3. Reading and Writing?**

Reading and Writing.

## Educational Insights

**Q1. Major patterns?**
Scores rise with parental education, standard lunch beats free/reduced, test prep helps, and subjects are highly correlated.

**Q2. Which demographic factors show differences?**
Lunch type and parental education most clearly, then gender and race/ethnicity.

**Q3. Patterns related to test prep?**
Completers score higher in every subject, most in Writing.

**Q4. What needs further investigation?**
Why socioeconomic indicators track so closely with scores, the gender gaps by subject, and the lower-scoring groups.

**Q5. Recommendations?**
- Target support for free/reduced lunch students.
- Expand access to test prep.
