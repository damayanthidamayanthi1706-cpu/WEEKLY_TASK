```python
df[["price", "bedrooms", "bathrooms",
    "sqft_living"]].corr()
```

```python
df["Length of House"] = (
    df["sqft_above"] +
    df["sqft_basement"]
)

df[["sqft_above", "sqft_basement",
    "Length of House"]].head()
```

```python
df["Built Year"] = df["yr_built"]

df[["yr_built", "Built Year"]].head()
```

```python
df["Sale Month"] = df["date"].dt.month_name()

df[["date", "Sale Month"]].head()
```

```python
df.groupby(df["date"].dt.year)["price"].mean()
```

```python
monthly_sales = (
    df.groupby("Sale Month")["id"]
    .count()
    .reset_index()
)

monthly_sales
```

```python
print(type(plt))
```

```python
plt.figure(figsize=(9,5))

sns.lineplot(
    data=monthly_sales,
    x="Sale Month",
    y="id",
    marker="o"
)

plt.title("Monthly House Sales")
plt.xlabel("Sale Month")
plt.ylabel("Number of Houses Sold")

plt.show()
```
